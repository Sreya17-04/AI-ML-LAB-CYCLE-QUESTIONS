```python
import heapq

graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('F', 3)],
    'D': [('G', 1)],
    'E': [('G', 2)],
    'F': [('G', 2)],
    'G': []
}

heuristic = {
    'A': 7,
    'B': 6,
    'C': 2,
    'D': 1,
    'E': 3,
    'F': 1,
    'G': 0
}

def astar(start, goal):
    queue = [(heuristic[start], 0, start, [start])]
    visited = set()

    while queue:
        total, cost, current, path = heapq.heappop(queue)

        if current == goal:
            return cost, path

        if current in visited:
            continue

        visited.add(current)

        for next, weight in graph[current]:
            new_cost = cost + weight
            total = new_cost + heuristic[next]
            heapq.heappush(queue, (total, new_cost, next, path + [next]))

cost, path = astar('A', 'G')

print("Cost:", cost)
print("Path:", path)
```
